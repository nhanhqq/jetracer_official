# OpenCV Road Following Live (No ML Model)

Notebook này sử dụng thuật toán Computer Vision truyền thống (Canny Edge Detection + Hough Transform) của OpenCV để tìm vạch kẻ đường thay vì dùng mạng nơ-ron.

Khởi tạo Camera và Xe (Sử dụng `JetRacerController` từ `basic_motion.py`)

```python
import os
import time
import cv2
import numpy as np
import ipywidgets
import threading
from IPython.display import display
from jetcam.csi_camera import CSICamera
from jetcam.utils import bgr8_to_jpeg
from basic_motion import JetRacerController

# Khởi động lại camera daemon
os.system('echo "jetson" | sudo -S systemctl restart nvargus-daemon')
time.sleep(2)

try:
    if 'camera' in globals():
        camera.running = False
        camera.unobserve_all()
except:
    pass

camera = CSICamera(width=224, height=224, capture_fps=0)

# Khởi tạo bộ điều khiển xe
car = JetRacerController()
```

Tạo giao diện hiển thị Live View

```python
state_widget = ipywidgets.ToggleButtons(options=['stop', 'live'], description='Trạng thái', value='stop')
prediction_widget = ipywidgets.Image(format='jpeg', width=camera.width, height=camera.height)
prediction_widget.value = bgr8_to_jpeg(np.zeros((224, 224, 3), dtype=np.uint8))

steering_gain_slider = ipywidgets.FloatSlider(description='Steering Gain', min=0.0, max=2.0, value=1.0, step=0.05, orientation='horizontal')
throttle_slider = ipywidgets.FloatSlider(description='Throttle', min=0.0, max=0.5, value=0.15, step=0.01, orientation='horizontal')

ui_widget = ipywidgets.VBox([
    prediction_widget,
    ipywidgets.HBox([state_widget]),
    ipywidgets.HBox([steering_gain_slider, throttle_slider])
])

display(ui_widget)
```

Thuật toán xử lý ảnh OpenCV để tìm đường đi và điều khiển xe

```python
def process_cv_lane(img):
    height, width = img.shape[:2]
    
    # 1. Cắt ROI (Region of Interest)
    roi_top = height // 2
    roi = img[roi_top:height, :]
    
    # 2. Xử lý lóe sáng / màu chập chờn bằng CLAHE (Contrast Limited Adaptive Histogram Equalization)
    lab = cv2.cvtColor(roi, cv2.COLOR_BGR2LAB)
    l, a, b = cv2.split(lab)
    clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8,8))
    cl = clahe.apply(l)
    limg = cv2.merge((cl,a,b))
    roi_clahe = cv2.cvtColor(limg, cv2.COLOR_LAB2BGR)
    
    # 3. Phân đoạn vật cản (Segmentation) tạo ảnh trắng đen
    gray = cv2.cvtColor(roi_clahe, cv2.COLOR_BGR2GRAY)
    blur = cv2.GaussianBlur(gray, (7, 7), 0)
    
    # Dùng Adaptive Threshold (tương tự thuật toán tách dải tần/hình thái) để tạo mask trắng đen
    thresh = cv2.adaptiveThreshold(blur, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C, cv2.THRESH_BINARY_INV, 11, 2)
    
    # Dọn nhiễu bằng Morphological Operations
    kernel = np.ones((5,5), np.uint8)
    thresh = cv2.morphologyEx(thresh, cv2.MORPH_OPEN, kernel)
    
    # Lọc bỏ vùng viền sa bàn ra khỏi vật cản (để không nhận vạch là vật)
    hsv = cv2.cvtColor(roi_clahe, cv2.COLOR_BGR2HSV)
    lower_orange = np.array([5, 100, 100])
    upper_orange = np.array([25, 255, 255])
    mask_orange = cv2.inRange(hsv, lower_orange, upper_orange)
    
    lower_red = np.array([0, 100, 100])
    upper_red = np.array([5, 255, 255])
    mask_red = cv2.inRange(hsv, lower_red, upper_red)
    
    mask_track = cv2.bitwise_or(mask_orange, mask_red)
    mask_track = cv2.dilate(mask_track, kernel, iterations=2)
    
    # Trừ phần vạch ra, ta có ảnh trắng đen chứa vật cản
    thresh_obs = cv2.bitwise_and(thresh, thresh, mask=cv2.bitwise_not(mask_track))
    
    # Tìm vùng vật cản
    contours_obs, _ = cv2.findContours(thresh_obs, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
    obstacle_detected = False
    obs_center_x = -1
    
    line_image = np.zeros_like(img)
    
    if contours_obs:
        contours_obs = sorted(contours_obs, key=cv2.contourArea, reverse=True)
        for cnt in contours_obs:
            bx, by, bw, bh = cv2.boundingRect(cnt)
            area = bw * bh
            # Chỉ lấy các khối có kích thước giống vật cản thực sự
            if 600 < area < 8000 and (bw/bh) < 3.0:
                obs_center_x = bx + bw // 2
                cv2.rectangle(line_image, (bx, by + roi_top), (bx+bw, by+bh + roi_top), (0, 255, 255), 2)
                cv2.putText(line_image, "Vat can", (bx, by + roi_top - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 255, 255), 2)
                obstacle_detected = True
                break
                
    # 4. Tìm các vạch đường (Canny + Hough Lines)
    edges = cv2.Canny(blur, 50, 150)
    lines = cv2.HoughLinesP(edges, 1, np.pi/180, threshold=30, minLineLength=15, maxLineGap=40)
    
    left_lines = []
    right_lines = []
    center_lines = []
    
    if lines is not None:
        for line in lines:
            x1, y1, x2, y2 = line[0]
            if x1 == x2: continue
            
            y1 += roi_top
            y2 += roi_top
            
            slope = (y2 - y1) / (x2 - x1)
            
            if slope < -0.3 and (x1 < width * 0.45 or x2 < width * 0.45):
                left_lines.append(line)
                cv2.line(line_image, (x1, y1), (x2, y2), (255, 0, 0), 3) # Vạch trái
            elif slope > 0.3 and (x1 > width * 0.55 or x2 > width * 0.55):
                right_lines.append(line)
                cv2.line(line_image, (x1, y1), (x2, y2), (0, 0, 255), 3) # Vạch phải
            elif abs(slope) > 0.5 and (width * 0.35 <= x1 <= width * 0.65):
                center_lines.append(line)
                cv2.line(line_image, (x1, y1), (x2, y2), (0, 255, 0), 3) # Vạch đứt ở giữa
                
    # 5. Tính điểm đích để bám đường
    target_x = width // 2
    target_y = height // 2 + 50
    
    left_x = []
    right_x = []
    mid_x = []
    
    if left_lines:
        for l in left_lines: left_x.extend([l[0][0], l[0][2]])
    if right_lines:
        for l in right_lines: right_x.extend([l[0][0], l[0][2]])
    if center_lines:
        for l in center_lines: mid_x.extend([l[0][0], l[0][2]])
        
    avg_left = sum(left_x)//len(left_x) if left_x else 0
    avg_right = sum(right_x)//len(right_x) if right_x else width
    avg_mid = sum(mid_x)//len(mid_x) if mid_x else -1
    
    if avg_mid != -1:
        # Nếu bắt được vạch đứt giữa -> Ưu tiên bám vạch giữa
        target_x = avg_mid
    elif left_lines and right_lines:
        target_x = (avg_left + avg_right) // 2
    elif left_lines:
        target_x = avg_left + 80
    elif right_lines:
        target_x = avg_right - 80
        
    # 6. Né vật cản (Override đích đến)
    if obstacle_detected:
        if obs_center_x < width // 2:
            target_x += 80 # Né phải
            cv2.putText(line_image, "Tranh Trai -> Re Phai", (50, 50), cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0, 255, 255), 2)
        else:
            target_x -= 80 # Né trái
            cv2.putText(line_image, "Tranh Phai -> Re Trai", (50, 50), cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0, 255, 255), 2)
            
    cv2.circle(line_image, (target_x, target_y), 8, (0, 255, 0), -1)
    result = cv2.addWeighted(img, 0.8, line_image, 1.0, 0)
    
    return result, target_x

def live_update(change):
    if state_widget.value != 'live':
        return
        
    img = change['new']
    processed_img, target_x = process_cv_lane(img.copy())
    
    steering = (target_x / (camera.width / 2.0)) - 1.0
    steering = steering * steering_gain_slider.value
    steering = max(min(steering, 1.0), -1.0)
    
    car.set_steering(steering)
    car.set_throttle(throttle_slider.value)
    
    prediction_widget.value = bgr8_to_jpeg(processed_img)

def state_changed(change):
    if change['new'] == 'stop':
        car.stop()

state_widget.observe(state_changed, names='value')
```

Chạy Camera

```python
camera.observe(live_update, names='value')
camera.running = True
```
